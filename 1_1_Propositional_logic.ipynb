# Building truth tables with Python

The truth table for $ p \wedge q $:

```python
print("p \t q \t p ∧ q \n")

for p in (True, False):
    for q in (True, False):
        print(p, "\t", q, "\t", p and q)
```

Truth table for $ p \wedge q \vee r $:

```python
print("p \t q \t r \t p ∧ q ∨ r \n")

for p in (True, False):
    for q in (True, False):
        for r in (True, False):
            print(p, "\t", q, "\t", r, "\t", p and q or r)
```

# Conditional statements

## The conditional statement $ p \rightarrow q $

```python
def cond(p, q):
    if p and not q:
        return False
    return True

print("p \t q \t p → q \n")
for p in (True, False):
    for q in (True, False):
        print(p, "\t", q, "\t", cond(p, q))
```

## The biconditional statement $ p \leftrightarrow q$

```python
print("p \t q \t p ↔ q \n")

for p in (True, False):
    for q in (True, False):
        print(p, "\t", q, "\t", cond(p, q) and cond(q, p)) # or simply p == q
```

# Operator forms

Mathematical operators come in three main forms:

1. Infix
2. Prefix
3. Postfix

An infix operator is most common:

```python
# operand, operator, operand
2 + 2
4 % 2
True and False
```

An example of a prefix operator is a function:

```python
def xor(p, q):
    return p ^ q

xor(False, False)
```

Or the $\neg$ operator:

```python
not True
```

An example of a postfix operator (sort of) is a method:

```python
"Hello, world".upper()
```

Some languages use *reverse Polish notation* (Wikipedia has examples), as in:

```python
# 2 2 + => 4
```

# Operator precedence

Operator precedence determines the order of evaluation of a series of operators making up a proposition; for example:

$$ p \vee q \wedge \neg r$$

cannot be correctly evaluated if just read from left to right. Operator precedence shows which operators to evaluate in which order. The correct order is

$$ \neg,  \wedge,  \vee,  \rightarrow,  \leftrightarrow. $$

```python
p = True
q = False
r = False

print(not r)
print(q and not r)
print(p or q and not r)
```
